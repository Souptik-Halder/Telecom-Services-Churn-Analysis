```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('WA_Fn-UseC_-Telco-Customer-Churn.csv')
df
```

```python
df.info()
```

```python
#REPLACING THE BLANKS WITH ZERO IN THE MONTHLY CHARGES COLUMN AND CHANGING ITS DATATYPE
```

```python
df["TotalCharges"]=df["TotalCharges"].replace(" ","0")
df["TotalCharges"]=df["TotalCharges"].astype(float)
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
df["customerID"].duplicated().sum()
```

```python
##CONVERTED 0s and 1s VALUE OF SENIOR CITIZEN INTO YES/NO TO MAKE IT UNDERSTANDABLE
```

```python
def conv(value):
    if value==1:
        return "yes"
    else:
        return "no"

df['SeniorCitizen']=df["SeniorCitizen"].apply(conv)
```

```python
df.head()
```

```python
ax=sns.countplot(x="Churn",data=df)
ax.bar_label(ax.containers[0])
plt.title("Count of customers by Chunk")
plt.show()
```

```python
gb=df.groupby("Churn").agg({"Churn":"count"})
plt.pie(gb["Churn"],labels=gb.index,autopct="%1.2f%%",shadow=True)
plt.title("% of churned customer")
plt.show()
```

```python
plt.figure(figsize=(5,6))
sns.countplot(x=df["gender"],data = df, hue="Churn")
plt.title("Churn by Gender")
plt.show()
```

```python
plt.figure(figsize=(5,6))
sns.countplot(x=df["SeniorCitizen"],data = df, hue="Churn")
plt.title("Churn by SeniorCitizen")
plt.show()
```

```python
##COMAPRATIVELY A GREATER AMOUNT(%) OF SENIOR CITIZENS HAVE CHURNED
```

```python
plt.figure(figsize=(9,5))
sns.histplot(x="tenure",data=df,bins=50,hue="Churn")
plt.show()
```

```python
#PEOPLE WHO HAVE USED OUR SERVICES FOR A LONG TIME HAVE STAYED AND PEOPLE WHO USED OUR SERVICES FOR ONE OR TWO MONTHS HAVE CHURNED
```

```python
plt.figure(figsize=(5,6))
ax=sns.countplot(x=df["Contract"],data = df,hue="Churn")
ax.bar_label(ax.containers[0])
plt.title("Count of customers by Contract")
plt.show()
```

```python
#CUSTOMERS WHO HAVE MONTH TO MONTH CONTARCT ARE MORE LIKELY TO CHURN AS COMPARED TO CUSTOMERS WHO HAVE 1 or 2 YEAR CONTRACT
```

```python
df.columns.values
```

```python
columns = [
    'PhoneService', 'MultipleLines', 'InternetService',
    'OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
    'TechSupport', 'StreamingTV', 'StreamingMovies'
]

# Create subplots
fig, axes = plt.subplots(3, 3, figsize=(15, 12))
axes = axes.flatten()

# Generate countplots for each column
for i, column in enumerate(columns):
    sns.countplot(x=df[column],data=df, ax=axes[i], hue="Churn")
    axes[i].set_title(f'Countplot of {column}')
    axes[i].set_xlabel('')
    axes[i].set_ylabel('Count')

# Adjust layout
plt.tight_layout()
plt.show()
```

```python
#Summary of Observations from the Subplots:
#Phone Service & Multiple Lines:

#The majority of customers have a phone service.
#A smaller number of customers have multiple lines compared to single-line users.
#Customers without a phone service have very low churn rates.
#Internet Service:

#The dataset includes three categories: DSL, Fiber Optic, and No Internet.
#Fiber optic users have the highest churn rate, while DSL and No Internet users have lower churn.
#Online Security & Online Backup:

#Many customers do not subscribe to these services.
#Churn is noticeably higher among customers who do not have these services, suggesting they may add value to customer retention.
#Device Protection & Tech Support:

#Similar patterns as Online Security and Backup.
#Customers without these services tend to have higher churn rates.
#Streaming TV & Streaming Movies:

#A significant portion of customers do not use these services.
#The churn rate is similar for both Streaming TV and Streaming Movies users, but customers without these services appear to churn more often.
#Key Insights:
#Customers without additional services (security, backup, tech support) are more likely to churn.
#Fiber optic users have a higher churn rate, which might indicate dissatisfaction.
#No phone service customers have very low churn, possibly due to a niche user base.
```
